# Clusterización

La intención de este documento es clasificar los archivos en 3 tipo de clima, soleado, nublado y lluvioso.
A pesar de que ya existe una columna con el tipo de cielo, en modo de réplica de los artículos aplicaremos técnicas de clusterización para dividir la información.

## K-Means

En este apartado vamos a usar K-Means para 3 de los 4 artículos y los datos de la técnica propuesta, es decir, modificaremos el archivo base y el de Photovoltaic.

### Base

```python
import pandas as pd
datos = pd.read_excel('02.Generación Pearson_2.xlsx')
datos.dtypes
```

```python
#Importaciones
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import MinMaxScaler
```

```python
# Extraer la hora de la columna 'Fecha'
datos["Hora"] = datos["Fecha"].dt.hour

# Filtrar los datos entre 6 AM (06:00) y 7 PM (19:00)
datos_filtrados = datos[(datos["Hora"] >= 6) & (datos["Hora"] <= 21)]

# Eliminar la columna "Hora" si no es necesaria
datos_filtrados = datos_filtrados.drop(columns=["Hora"])

# Mostrar las primeras filas para verificar
datos_filtrados.head()
```

```python

datos_scaled = datos_filtrados[['Temperatura', 'Probabilidad Lluvia', 'Humedad Relativa', 'Índice UV', 'Condición Cielo', 'DPT']].copy()
datos_scaled
```

```python
#Estandarizamos por min max
scaler = MinMaxScaler()
scaler
```

```python
col_num = datos_scaled[['Temperatura', 'Probabilidad Lluvia', 'Humedad Relativa', 'Índice UV', 'Condición Cielo', 'DPT']].columns
datos_scaled[col_num] = scaler.fit_transform(datos_scaled[col_num])
datos_scaled
```

```python
pip install yellowbrick
```

```python

from sklearn.cluster import KMeans
from yellowbrick.cluster import KElbowVisualizer
# Determinar el número óptimo de clusters usando el método del codo
model = KMeans(n_init=10)
visualizer = KElbowVisualizer(model, k=(2,10), metric='distortion', timings=False)
visualizer.fit(datos_scaled)
visualizer.show()
```

```python
from sklearn.metrics import silhouette_score

k_values = range(2, 11)
silhouettes = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(datos_scaled)
    score = silhouette_score(datos_scaled, labels)
    silhouettes.append(score)
```

```python
plt.figure(figsize=(8,5))
plt.plot(k_values, silhouettes, marker = 'o', linestyle = '-')
plt.xlabel("Número de Clusters (k)")
plt.ylabel("Coeficiente de Silhouette")
plt.xticks(k_values)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Aplicar PCA para reducir a 2 dimensiones **antes** de K-Means
pca = PCA(n_components=2)
datos_pca = pca.fit_transform(datos_scaled)

# Aplicar K-Means sobre los datos transformados con PCA
k = 5  # Puedes ajustar este valor según el análisis previo
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
clusters = kmeans.fit_predict(datos_pca)

# Crear un DataFrame con los componentes principales y los clusters asignados
df_pca = pd.DataFrame(datos_pca, columns=["PC1", "PC2"])
df_pca["Cluster"] = clusters

# Obtener los centroides directamente desde K-Means (ahora en el espacio PCA)
centroids_pca = kmeans.cluster_centers_

# Visualizar los clusters en un gráfico de dispersión
plt.figure(figsize=(8,6))
for cluster in range(k):
    plt.scatter(df_pca[df_pca["Cluster"] == cluster]["PC1"],
                df_pca[df_pca["Cluster"] == cluster]["PC2"],
                label=f'Cluster {cluster}')

# Graficar los centroides
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1], 
            marker="X", s=200, c="black", label="Centroides")

plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.title("Clusters visualizados en 2D con PCA")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import calinski_harabasz_score
ch_scores = []
k_values = range(2, 11)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(datos_scaled)
    ch_scores.append(calinski_harabasz_score(datos_scaled, clusters))

# Graficar CH Index vs número de clusters
plt.figure(figsize=(8,5))
plt.plot(k_values, ch_scores, marker='o', linestyle='-')
plt.xlabel("Número de Clusters (k)", fontsize=12)
plt.ylabel("Índice de Calinski-Harabasz", fontsize=12)
plt.grid(True)
plt.show()
```

## Gaussian Mixture Model

```python
datos_GMM = datos.copy()
```

```python
# Filtrar los datos entre 6 AM (06:00) y 7 PM (19:00)
datos_filtrados_GMM = datos_GMM[(datos_GMM["Hora"] >= 6) & (datos_GMM["Hora"] <= 21)]

# Eliminar la columna "Hora" si no es necesaria
datos_filtrados_GMM = datos_filtrados_GMM.drop(columns=["Hora"])

# Mostrar las primeras filas para verificar
datos_filtrados_GMM.head()
```

```python

datos_scaled_GMM = datos_filtrados_GMM[['Temperatura', 'Probabilidad Lluvia', 'Humedad Relativa', 'Índice UV', 'Condición Cielo', 'DPT']].copy()
datos_scaled_GMM
```

```python
#Estandarizamos por min max
scaler_GMM = MinMaxScaler()
scaler_GMM
```

```python
col_num = datos_scaled_GMM[['Temperatura', 'Probabilidad Lluvia', 'Humedad Relativa', 'Índice UV', 'Condición Cielo', 'DPT']].columns
datos_scaled_GMM[col_num] = scaler_GMM.fit_transform(datos_scaled_GMM[col_num])
datos_scaled_GMM
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# Rango de valores de k (número de clusters)
k_values = range(2, 11)

# Listas para almacenar métricas
bic_scores = []
aic_scores = []
silhouette_scores = []
ch_scores = []

# Evaluar modelos GMM para diferentes valores de k
for k in k_values:
    gmm = GaussianMixture(n_components=k, random_state=42)
    gmm.fit(datos_scaled_GMM)
    
    # Calcular métricas
    bic_scores.append(gmm.bic(datos_scaled_GMM))
    aic_scores.append(gmm.aic(datos_scaled_GMM))
    
    # Asignar clusters y calcular Silhouette Score
    labels = gmm.predict(datos_scaled_GMM)
    silhouette_scores.append(silhouette_score(datos_scaled_GMM, labels))

    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(datos_scaled_GMM)
    ch_scores.append(calinski_harabasz_score(datos_scaled_GMM, clusters))


```

```python
# Graficar los resultados
plt.figure(figsize=(20, 6))

size = 20

# Gráfico de BIC
plt.subplot(1, 3, 1)
plt.plot(k_values, np.array(bic_scores)/1000, marker='o', linestyle='-', color="#579de4")
plt.xlabel("Número de Clusters (k)", fontsize=size)
plt.ylabel("Índice BIC (×10³)", fontsize=size)
plt.yticks(fontsize=size)
plt.xticks(fontsize=size)
plt.grid(True, which='both', axis='both', linestyle='--', linewidth=0.5, color='gray', alpha=0.3)

# Gráfico de Silhouette Score
plt.subplot(1, 3, 2)
plt.plot(k_values, silhouette_scores, marker='o', linestyle='-', color="#579de4")
plt.xlabel("Número de Clusters (k)", fontsize=size)
plt.ylabel("Índice Silhouette", fontsize=size)
plt.yticks(fontsize=size)
plt.xticks(fontsize=size)
plt.grid(True, which='both', axis='both', linestyle='--', linewidth=0.5, color='gray', alpha=0.3)

# Gráfico de Silhouette Score
plt.subplot(1, 3, 3)
plt.plot(k_values, np.array(ch_scores) / 1000, marker='o', linestyle='-', color="#579de4")
plt.xlabel("Número de Clusters (k)", fontsize=size)
plt.ylabel("Índice Calinski-Harabasz (×10³)", fontsize=size)
plt.yticks(fontsize=size)
plt.xticks(fontsize=size)
plt.grid(True, which='both', axis='both', linestyle='--', linewidth=0.5, color='gray', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Aplicar GMM con el número óptimo de clusters
k_optimo = 3
gmm_final = GaussianMixture(n_components=k_optimo, random_state=42)
clusters_gmm = gmm_final.fit_predict(datos_scaled_GMM)

# Asignar los clusters al DataFrame original
datos_scaled_GMM["Cluster"] = clusters_gmm
datos_scaled_GMM
```
